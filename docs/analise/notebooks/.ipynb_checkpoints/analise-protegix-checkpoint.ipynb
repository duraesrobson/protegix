# Análise dos Dados — Protegix

> Pesquisa sobre segurança digital, LGPD e IA
> Amostragem: 101 participantes

---
## Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="pastel")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["figure.dpi"] = 120

print("Setup completo ✅")
```

---
## 1. Carregar dados

```python
agregado = pd.read_csv("../../dados/brutos/respostas-agregadas.csv", sep=";")
confianca = pd.read_csv("../../dados/tratados/escalas-confianca.csv", sep=";")
indices = pd.read_csv("../../dados/tratados/indices-gerais.csv", sep=";")

print(f"Agregado: {agregado.shape[0]} linhas")
print(f"Confiança: {confianca.shape[0]} linhas")
print(f"Índices: {indices.shape[0]} linhas")
```

```python
agregado.head(10)
```

---
## 2. Perfil dos Participantes

```python
perfil_ids = ["faixa_etaria", "escolaridade", "tempo_uso", "tipo_conexao"]
perfil = agregado[agregado["pergunta_id"].isin(perfil_ids)]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for ax, pid in zip(axes.flat, perfil_ids):
    df = perfil[perfil["pergunta_id"] == pid]
    ax.barh(df["resposta"], df["percentual"])
    ax.set_title(pid.replace("_", " ").title())
    ax.set_xlabel("%")

plt.tight_layout()
plt.show()
```

---
## 3. Segurança Digital

```python
seg_ids = ["senhas_diferentes", "2fa", "golpe", "verifica_sites", "disp_atualizacao"]
seg = agregado[agregado["pergunta_id"].isin(seg_ids)]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flat

for i, pid in enumerate(seg_ids):
    df = seg[seg["pergunta_id"] == pid]
    cores = sns.color_palette("RdYlGn_r", len(df))
    axes[i].pie(df["percentual"], labels=df["resposta"], autopct="%1.0f%%",
                colors=cores, startangle=90)
    axes[i].set_title(pid.replace("_", " ").title())

axes[5].remove()
plt.tight_layout()
plt.show()
```

---
## 4. Níveis de Confiança (0–100)

```python
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=confianca, x="contexto", y="media_0_100",
                 palette="viridis")
ax.bar_label(ax.containers[0], fmt="%.0f")
plt.title("Confiança por Contexto")
plt.ylabel("Média (0–100)")
plt.xlabel("")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

---
## 5. LGPD e Privacidade

```python
lgpd_ids = ["lgpd", "lgpd_nivel", "lgpd_termos", "lgpd_uso"]
lgpd = agregado[agregado["pergunta_id"].isin(lgpd_ids)]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for ax, pid in zip(axes.flat, lgpd_ids):
    df = lgpd[lgpd["pergunta_id"] == pid]
    ax.barh(df["resposta"], df["percentual"])
    ax.set_title(pid.replace("_", " ").title())
    ax.set_xlabel("%")

plt.tight_layout()
plt.show()
```

---
## 6. Inteligência Artificial

```python
ia_ids = ["utiliza_ia", "frequencia_ia", "finalidade_ia", "confianca_ia", "verificacao_ia"]
ia = agregado[agregado["pergunta_id"].isin(ia_ids)]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flat

for i, pid in enumerate(ia_ids):
    df = ia[ia["pergunta_id"] == pid]
    axes[i].pie(df["percentual"], labels=df["resposta"], autopct="%1.0f%%",
                startangle=90)
    axes[i].set_title(pid.replace("_", " ").title())

axes[5].remove()
plt.tight_layout()
plt.show()
```

---
## 7. Índices Compostos

```python
plt.figure(figsize=(12, 5))
sns.barplot(data=indices, x="dimensao", y="valor", hue="indicador")
plt.title("Indicadores por Dimensão")
plt.ylabel("%")
plt.xticks(rotation=10)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

---
## 8. Resumo Geral

- **Celular domina**: 88% acessam a internet pelo smartphone
- **Higiene digital frágil**: apenas 27% sempre diversificam senhas, 35% usam 2FA
- **IA popular mas com ceticismo**: 80% usam, mas só 12% confiam plenamente
- **LGPD conhecida mas não praticada**: 68% conhecem, 50% nunca leem termos
- **Confiança maior em bancos (64) que em navegação comum (45)**
